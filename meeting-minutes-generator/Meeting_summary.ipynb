```python
# imports

import os
import requests
from IPython.display import Markdown, display, update_display
from openai import OpenAI
from google.colab import drive
from huggingface_hub import login
from google.colab import userdata
from transformers import AutoTokenizer, AutoModelForCausalLM, TextStreamer, BitsAndBytesConfig
import torch
```

```python
# Constants

LLAMA = "meta-llama/Llama-3.2-3B-Instruct"
```

```python
drive.mount("/content/drive")
audio_filename = "/content/drive/MyDrive/llm/denver_extract.mp3"
```

```python
hf_token = userdata.get('HF_TOKEN')
login(hf_token, add_to_git_credential = True)

#open file

audio_file = open(audio_filename,"rb")
```

## Step 1: Transcribing audio

```python
from transformers import pipeline

pipe = pipeline(
    "automatic-speech-recognition",
    model="openai/whisper-medium.en",
    dtype=torch.float16,
    device='cuda',
    return_timestamps=True
)

result = pipe(audio_filename)
transcription = result["text"]
print(transcription)
```

## Step 2: Analyze & report

```python
system_message = """
You produce minutes of meeting from transcripts, with summary, key discussion points, takeaways and action items
with owners, in markdown format without code blocks.
"""

user_prompt = f"""
Below is an extract of a Denver city council meeting.
Please write minutes in markdown without code blocks, including:
- a summary with attendees, location and date
- discussion points
- takeaways
- action item with owners

Transcription:
{transcription}
"""

message = [
    {"role": "system", "content": system_message},
    {"role": "user", "content": user_prompt}
]
```

```python
# using BitsAndBytesConfig to quantise 4 bits

quant_config = BitsAndBytesConfig(
    load_in_4bit = True,
    bnb_4bit_use_double_quant = True,
    bnb_4bit_compute_dtype = torch.bfloat16,
    bnb_4bit_quant_type = "nf4"
)
```

```python
# Need LLAMA approval first

tokenizer = AutoTokenizer.from_pretrained(LLAMA)
tokenizer.pad_token = tokenizer.eos_token
inputs = tokenizer.apply_chat_template(message, return_tensors="pt").to("cuda")
streamer = TextStreamer(tokenizer)
model = AutoModelForCausalLM.from_pretrained(LLAMA, device_map = "auto", quantization_config = quant_config)
outputs = model.generate(inputs, max_new_tokens = 2000, streamer = streamer)
```

```python
response = tokenizer.decode(outputs[0])
```

```python
display(Markdown(response))
```
